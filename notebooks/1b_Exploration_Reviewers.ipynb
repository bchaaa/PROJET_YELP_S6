# Phase 1b : Analyse des Utilisateurs (Reviewers)
Objectif : Determiner si les utilisateurs experts (Elite) sont plus severes dans leur notation.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os
import sys
from scipy.stats import ttest_ind
sys.path.append(os.path.abspath('..'))
from src.data_loader import load_yelp_sample
sns.set_theme()
```

```python
# Chargement des donnees Utilisateurs
df_user = load_yelp_sample('../data/raw/user.json', n_rows=50000)
```

```python
# Creation d'une colonne binaire pour identifier les membres Elite
df_user['is_elite'] = df_user['elite'].apply(lambda x: 1 if x != '' and x != 'None' else 0)

# Separation des notes moyennes des deux groupes
elite = df_user[df_user['is_elite'] == 1]['average_stars']
norm = df_user[df_user['is_elite'] == 0]['average_stars']

print(f'Note moyenne Elite : {elite.mean():.2f}')
print(f'Note moyenne Standard : {norm.mean():.2f}')

# Affichage de la densite de probabilite (kde)
plt.figure(figsize=(8, 6))
sns.kdeplot(elite, label='Membres Elite', fill=True)
sns.kdeplot(norm, label='Utilisateurs Standards', fill=True)
plt.title('Distribution des notes moyennes par type d\'utilisateur')
plt.legend()
plt.show()
```
