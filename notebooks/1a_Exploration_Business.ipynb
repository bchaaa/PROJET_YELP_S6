# Phase 1a : Analyse des Commerces (Business)
Objectif : Verifier si les etablissements les plus frequentes ont de meilleures notes que les autres.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os
import sys
from scipy.stats import mannwhitneyu
sys.path.append(os.path.abspath('..'))
from src.data_loader import load_yelp_sample
# Configuration du style des graphiques
sns.set_theme()
```

```python
# Chargement des donnees Business (50 000 lignes)
df_biz = load_yelp_sample('../data/raw/business.json', n_rows=50000)
```

```python
# On separe les commerces en deux groupes : les tres populaires (top 10%) et les autres
threshold = df_biz['review_count'].quantile(0.9)
pop = df_biz[df_biz['review_count'] > threshold]['stars']
other = df_biz[df_biz['review_count'] <= threshold]['stars']

# Test statistique de Mann-Whitney U pour comparer les distributions
stat, p = mannwhitneyu(pop, other)
print(f'P-value du test : {p:.2e}')

# Visualisation sous forme de boites a moustaches
plt.figure(figsize=(8, 6))
sns.boxplot(data=[pop, other], palette='Set2').set_xticklabels(['Populaires', 'Standards'])
plt.title('Comparaison des notes : Populaires vs Standards')
plt.show()
```
